# Exploratory strategy model

This notebook demonstrates loading prices and comparing normal and volatility-adjusted momentum.

```python
from pathlib import Path
import sys

ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
SRC = ROOT / "src"
if str(SRC) not in sys.path:
    sys.path.insert(0, str(SRC))
```

```python
import backtest
from data import load_prices

tickers = ["URTH", "GLD", "IEF", "SHV"]
prices = load_prices(tickers)
normal = backtest.run(prices, lookback=12, skip=1, common=True)
volatility_adjusted = backtest.run(
    prices, lookback=12, skip=1, common=True, volatility_adjusted=True
)
```

```python
import plotly.express as px

comparison = {
    "Normal": normal.equity,
    "Volatility-adjusted": volatility_adjusted.equity,
}
px.line(comparison, title="Momentum strategy comparison")
```
